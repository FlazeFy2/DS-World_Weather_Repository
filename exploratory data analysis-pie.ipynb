```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("GlobalWeatherRepository.csv")

# Exploratory Data Analysis (EDA) - Pie Chart Most Context for Nonconditional Data
target_col = ['condition_text','wind_direction','moon_phase']

for dt in target_col:
    # Dataset
    top_context = df[dt].value_counts().head(7)

    # Pie Chart
    plt.figure(figsize=(8,8))
    top_context.plot.pie(autopct="%1.1f%%", startangle=90)
    plt.title(f"Most {dt}")
    plt.ylabel("")
    plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("GlobalWeatherRepository.csv")

# Exploratory Data Analysis (EDA) - Pie Chart Humidity Category Comparison
def categorize_humid(val):
    if val > 80:
        return "Very High (>80)"
    elif val >= 50:
        return "High (50-80)"
    else:
        return "Low (<50)"

df["Humidity Category"] = df["humidity"].apply(categorize_humid)

# Dataset
top_context = df["Humidity Category"].value_counts()

# Pie Chart
plt.figure(figsize=(8,8))
top_context.plot.pie(autopct="%1.1f%%", startangle=90)
plt.title("Humidity Category Comparison")
plt.ylabel("")
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("GlobalWeatherRepository.csv")

# Exploratory Data Analysis (EDA) - Pie Chart PM2 Air Quality Category Comparison
def categorize_humid(val):
    if val > 250.5:
        return "Hazardous"
    elif val >= 150.5:
        return "Very Unhealthy"
    elif val >= 55.5:
        return "Unhealthy"
    elif val >= 35.5:
        return "Unhealthy for Sensitive Groups"
    elif val >= 12.1:
        return "Moderate"
    else:
        return "Good"

df["PM2 Air Quality Category"] = df["air_quality_PM2.5"].apply(categorize_humid)

# Dataset
top_context = df["PM2 Air Quality Category"].value_counts()

# Pie Chart without on-canvas labels
plt.figure(figsize=(8,8))
wedges, texts, autotexts = plt.pie(
    top_context, labels=None, autopct="%1.1f%%", startangle=90, wedgeprops={'linewidth': 1, 'edgecolor': 'white'}
)
plt.title("PM2 Air Quality Category Comparison")

# Legend at bottom
plt.legend(wedges,top_context.index,title="Category",
    loc="lower center",bbox_to_anchor=(0.5, -0.1),ncol=2,frameon=False
)
plt.tight_layout()
plt.ylabel("")
plt.show()
```
